```python
import pandas as pd

df = pd.read_csv('Pizzas.csv')
df
```

```python
df.plot(kind="scatter", x="diametro", y="preco");
```

```python
from sklearn import linear_model

modelo = linear_model.LinearRegression() #Cria o modelo
x = df[["diametro"]]
y = df[["preco"]]

modelo.fit(x, y) #Treina o modelo
```

```python
print(modelo.predict([[23]])[0][0])
```
